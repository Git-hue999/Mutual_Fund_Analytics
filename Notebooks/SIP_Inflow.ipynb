```python

# DAY 3 - TASK 3: SIP INFLOW TREND

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Loading SIP data:
sip_df = pd.read_csv("../Data/Raw/04_monthly_sip_inflows.csv")

# Checking column names:
print("Columns:", sip_df.columns.tolist())

print("\nFirst 5 rows:")
display(sip_df.head())

# Clean column names:
sip_df.columns = (
    sip_df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

# Display cleaned data:
print("\nCleaned SIP data:")
display(sip_df.head())
```

```python
# DAY 3 - TASK 3: SIP INFLOW TREND

import pandas as pd
import matplotlib.pyplot as plt

# Convert month to date:
sip_df["month"] = pd.to_datetime(sip_df["month"])

# Sort by month:
sip_df = sip_df.sort_values("month")

# Create chart:
plt.figure(figsize=(15, 7))

plt.plot(
    sip_df["month"],
    sip_df["sip_inflow_crore"],
    linewidth=2
)

# Mark December 2025 milestone:
dec_2025 = sip_df[sip_df["month"] == "2025-12-01"]

if not dec_2025.empty:
    value = dec_2025["sip_inflow_crore"].iloc[0]

    plt.scatter(
        dec_2025["month"],
        value,
        s=80,
        zorder=5
    )

    plt.annotate(
        f"Dec 2025: ₹{value:,.0f} Cr",
        xy=(dec_2025["month"].iloc[0], value),
        xytext=(-120, 30),
        textcoords="offset points",
        arrowprops=dict(arrowstyle="->")
    )

plt.title("Monthly SIP Inflow Trend (2022–2025)", fontsize=16)
plt.xlabel("Month")
plt.ylabel("SIP Inflow (₹ Crore)")
plt.grid(True, alpha=0.3)
plt.tight_layout()

# Save chart:
plt.savefig(
    "../Reports/SIP_Inflow_Trend.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("SIP Inflow chart created successfully.")
print("Saved to: Reports/SIP_Inflow_Trend.png")
```
