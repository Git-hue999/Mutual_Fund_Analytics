# DAY 3 - TASK 10: EDA FINDINGS

## Key Exploratory Data Analysis Findings:

1. **NAV Recovery and Growth:**
   The NAV trend shows a clear recovery and upward movement across the 2022–2026 period, indicating improving fund performance after the COVID-related market disruption.

2. **Strong AUM Growth Across Major Fund Houses:**
   AUM increased consistently between 2022 and 2025 across the major fund houses. SBI Mutual Fund recorded the highest AUM among the displayed fund houses, reaching approximately ₹12.5 crore in 2025.

3. **SBI Mutual Fund Maintains Market Leadership:**
   SBI Mutual Fund remained the strongest fund house in terms of AUM throughout the 2022–2025 period, with its AUM increasing substantially over the four-year period.

4. **SIP Inflows Show Overall Positive Growth:**
   Monthly SIP inflows demonstrate an overall upward trend during the 2022–2025 period, indicating increasing investor participation through systematic investment plans.

5. **SIP Activity Indicates Growing Investor Participation:**
   The increase in active SIP accounts and new SIP accounts suggests that systematic investment products are gaining popularity among retail investors.

6. **Banking Has the Largest Portfolio Sector Allocation:**
   Banking represents the largest share of the top holdings sector distribution at approximately 22%, followed by IT at 15.3% and Pharma at 13.7%.

7. **Technology and Healthcare Remain Important Allocations:**
   IT and Pharma together account for approximately 29% of the illustrated sector allocation, showing significant exposure to technology and healthcare-related companies.

8. **Folio Count Nearly Doubled:**
   The folio count increased from 13.26 crore in January 2022 to 26.12 crore in December 2025, representing a substantial expansion in the number of mutual fund investor accounts.

9. **Strongest Folio Growth Occurred During 2024–2025:**
   The folio-count trend shows particularly strong acceleration from 2024 into 2025, indicating increasing participation in mutual fund investments during this period.

10. **Diversification Across Multiple Economic Sectors:**
    The sector allocation is distributed across Banking, IT, Pharma, Automobile, Utilities, FMCG, Infrastructure, Diversified, Telecom and Consumer Goods, indicating exposure to a broad range of economic sectors.
